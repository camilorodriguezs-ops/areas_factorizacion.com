### List JSON files

Let's start by listing all files ending with `.json` in your current directory to see what's available.

```python
import glob
import os

json_files = glob.glob('*.json')

if json_files:
    print('Found JSON files:')
    for f in json_files:
        print(f'- ' + f)
else:
    print('No JSON files found in the current directory.')
```

### Display the content of a JSON file

Once you know the name of the JSON file you want to inspect, you can use the following code to load and display its content. Replace `your_file_name.json` with the actual name of your JSON file.



```python
import json

# Reemplaza 'your_file_name.json' con el nombre real de tu archivo JSON.
file_name = 'proyectos_informaticos.empresa.json' # @param {type:"string"}

try:
    with open(file_name, 'r') as f:
        data = json.load(f)
    print(f'Content of {file_name}:')
    display(data)
except FileNotFoundError:
    print(f'Error: File "{file_name}" not found. Please make sure the file name is correct.')
except json.JSONDecodeError:
    print(f'Error: Could not decode JSON from "{file_name}". The file might be corrupted or not a valid JSON.')
except Exception as e:
    print(f'An unexpected error occurred: {e}')
```
